```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
# Load unemployment dataset

url = "https://raw.githubusercontent.com/Yash22222/Oasis-Infobyte-Data-Science-Internship/main/Unemployment_Rate_upto_11_2020.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
# Check dataset information

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDataset Information:")
df.info()
```

```python
# Display column names and missing values

print("Column Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Clean column names by removing extra spaces

df.columns = df.columns.str.strip()

print("Cleaned Column Names:")
print(df.columns.tolist())
```

```python
# Convert Date column into datetime format

df['Date'] = pd.to_datetime(df['Date'].str.strip(), dayfirst=True)

print("Date column converted successfully!")
print(df['Date'].head())
```

```python
# Unemployment Rate over time

plt.figure(figsize=(10,5))

plt.plot(
    df['Date'],
    df['Estimated Unemployment Rate (%)'],
    marker='o'
)

plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.title('Unemployment Rate Over Time')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Average unemployment rate by region

region_unemployment = df.groupby('Region')[
    'Estimated Unemployment Rate (%)'
].mean().sort_values(ascending=False)

plt.figure(figsize=(12,6))

region_unemployment.plot(kind='bar')

plt.xlabel('Region')
plt.ylabel('Average Unemployment Rate (%)')
plt.title('Average Unemployment Rate by Region')

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# Average unemployment rate by month

df['Month'] = df['Date'].dt.month

monthly_unemployment = df.groupby('Month')[
    'Estimated Unemployment Rate (%)'
].mean()

plt.figure(figsize=(8,5))

monthly_unemployment.plot(kind='bar')

plt.xlabel('Month')
plt.ylabel('Average Unemployment Rate (%)')
plt.title('Monthly Average Unemployment Rate')

plt.xticks(
    range(12),
    ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
     'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    rotation=0
)

plt.show()
```

```python
# Distribution of unemployment rate

plt.figure(figsize=(8,5))

sns.histplot(
    df['Estimated Unemployment Rate (%)'],
    bins=20,
    kde=True
)

plt.xlabel('Unemployment Rate (%)')
plt.ylabel('Frequency')
plt.title('Distribution of Unemployment Rate')

plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(
    df.select_dtypes(include='number').corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
print("Unemployment Rate Summary:")
print(df['Estimated Unemployment Rate (%)'].describe())
```

The unemployment analysis was performed using the dataset up to November 2020. The analysis included unemployment trends over time, average unemployment rates by region, monthly averages, distribution, and correlation between numerical variables.

The summary statistics show that the average estimated unemployment rate was approximately 12.24%, with a minimum of 0.50% and a maximum of 75.85%. The visualizations help identify changes in unemployment rates across different months and regions.
